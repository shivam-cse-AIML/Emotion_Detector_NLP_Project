```python
df['text'] = df['text'].apply(lambda x : x.lower())
```

```python
unique_emotion = df['emotion'].unique()
emotion_number = {}
i =0
for  emo in unique_emotion:
  emotion_number[emo] = i
  i = i+1
df['emotion'] = df['emotion'].map(emotion_number)
```

```python
df['text'] = df['text'].apply(lambda x : x.lower())
```

```python
import string
def remove_punc(txt):
  return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df.head()
```

```python
df['text'] = df['text'].apply(remove_punc)
```

```python
def remove_numbers(txt):
   new = ""
   for i in txt:
       if not i.isdigit():
           new = new + i
   return  new
df['text'] = df['text'].apply(remove_numbers)
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
```

```python
df.loc[1]['text']
```

```python
def remove(txt):
  words = txt.split()
  cleaned = []
  for i in words:
    if i  not in stop_words:
      cleaned.append(i)

  return ' '.join(cleaned)
```

```python
df['text'] =df['text'].apply(remove)
```

```python
from sklearn.linear_model import SGDClassifier
from sklearn.metrics import accuracy_score, f1_score, classification_report

sgd_model = SGDClassifier(
    loss='hinge',
    alpha=0.0001,
    max_iter=1000,
    random_state=42
)

sgd_model.fit(x_train_tfidf, y_train)

sgd_pred = sgd_model.predict(x_test_tfidf)

print("Accuracy:", accuracy_score(y_test, sgd_pred))
print("Macro F1:", f1_score(y_test, sgd_pred, average='macro'))

print(classification_report(y_test, sgd_pred))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import SGDClassifier
```

```python
model_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer()),
    ('sgd', SGDClassifier(
        loss='hinge',
        alpha=0.0001,
        max_iter=1000,
        random_state=42
    ))
])
```

```python
model_pipeline.fit(X_train, y_train)
```

```python
pipeline_pred = model_pipeline.predict(X_test)

print("Pipeline Accuracy:",
      accuracy_score(y_test, pipeline_pred))

print("Pipeline Macro F1:",
      f1_score(y_test, pipeline_pred, average='macro'))

print(classification_report(y_test, pipeline_pred))
```
